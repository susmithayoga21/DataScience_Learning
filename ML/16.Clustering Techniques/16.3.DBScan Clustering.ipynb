```python
import pandas as pd 
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN

X,y =make_moons( shuffle=True,random_state=39)
```

```python
moon_data = pd.DataFrame(X,columns=["X1","X2"])
moon_data["class"] = y 
```

```python
moon_data
```

```python
import matplotlib.pyplot as plt
plt.scatter(moon_data["X1"],moon_data["X2"])
```

#### 1. AgglomerativeClustering

```python
h_clustering = AgglomerativeClustering(linkage='single')
h_clustering.fit_predict(X)
moon_data["class_pr"] = h_clustering.labels_
```

```python
moon_data.head(5)
```

```python
plt.scatter(moon_data["X1"],moon_data["X2"],c=moon_data["class_pr"])
```

#### 2. Kmeans

```python
k_clustering = KMeans(n_clusters = 2)
k_clustering.fit_predict(X)
moon_data["class_pr"]  = k_clustering.labels_
```

```python
plt.scatter(moon_data["X1"],moon_data["X2"],c=moon_data["class_pr"])
```

#### 3. DBSCAN

```python
db_clustering = DBSCAN()
db_clustering.fit_predict(X)
moon_data["class_pr"]  =db_clustering.labels_
```

```python
plt.scatter(moon_data["X1"],moon_data["X2"],c=moon_data["class_pr"])
```
