### 2. Data Collection

#### 2.1 Import necessory libirary and datasets

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import GradientBoostingClassifier,AdaBoostClassifier

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import classification_report,confusion_matrix,accuracy_score,roc_auc_score

cc_data = pd.read_csv("credit_card_clean.csv")
```

```python
cc_data_copy01 = cc_data.copy()
```

#### 3.2 Data Transformation 

```python
encoder = LabelEncoder()
cc_data_copy01["SEX"] = encoder.fit_transform(cc_data_copy01["SEX"])
cc_data_copy01["EDUCATION"] = encoder.fit_transform(cc_data_copy01["EDUCATION"])
cc_data_copy01["MARRIAGE"] = encoder.fit_transform(cc_data_copy01["MARRIAGE"])
```

```python
cc_data_copy01.dtypes
```

### 4. Data Preparation

```python
X = cc_data_copy01.drop('DEFAULT',axis = 1) #cc_data_copy01.iloc[:,0:23]
y = cc_data_copy01['DEFAULT']
```

```python
 X_train, X_test, y_train, y_test  = train_test_split(X,y,train_size=0.8,
                                                      random_state=39,
                                                      shuffle=True,
                                                      stratify=y)
```

### 5. Model Building

```python
ada_boost = AdaBoostClassifier()
grad_boost = GradientBoostingClassifier(n_estimators=50)
xg_boost = XGBClassifier(n_estimators=50)
lg_boost = LGBMClassifier(n_estimators=50)
```

### 6. Model Training

```python
%%time
ada_boost.fit(X_train,y_train)
```

```python
%%time
grad_boost.fit(X_train,y_train)
```

```python
%%time
xg_boost.fit(X_train,y_train)
```

```python
%%time
lg_boost.fit(X_train,y_train)
```

### 7. Model Testing

```python
y_pred_adb = ada_boost.predict(X_test)
y_pred_gb = grad_boost.predict(X_test)
y_pred_xgb = xg_boost.predict(X_test)
y_pred_lgb = lg_boost.predict(X_test)
```

### 8. Model Evaluation

```python
print("ADA BOOST")
print(confusion_matrix(y_test,y_pred_adb))
print(classification_report(y_test,y_pred_adb))
print("Accuracy score",accuracy_score(y_test,y_pred_adb))
print("ROC AUC score",roc_auc_score(y_test,y_pred_adb))
```

```python
print("GRADIENT BOOST")
print(confusion_matrix(y_test,y_pred_gb))
print(classification_report(y_test,y_pred_gb))
print("Accuracy score",accuracy_score(y_test,y_pred_gb))
print("ROC AUC score",roc_auc_score(y_test,y_pred_gb))
```

```python
print("Extreme GRADIENT BOOST")
print(confusion_matrix(y_test,y_pred_xgb))
print(classification_report(y_test,y_pred_xgb))
print(accuracy_score(y_test,y_pred_xgb))
print(roc_auc_score(y_test,y_pred_xgb))
```

```python
print("Light GRADIENT BOOST")
print(confusion_matrix(y_test,y_pred_lgb))
print(classification_report(y_test,y_pred_lgb))
print(accuracy_score(y_test,y_pred_lgb))
print(roc_auc_score(y_test,y_pred_lgb))
```
