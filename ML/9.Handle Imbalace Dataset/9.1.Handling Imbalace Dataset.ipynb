### 2. Data Collection

#### 2.1 Import Necessory Libiraries and datasets

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
breast_cancer_data = load_breast_cancer()
```

```python
breast_cancer =pd.DataFrame(data = breast_cancer_data.data,columns = breast_cancer_data.feature_names)
breast_cancer['target'] = breast_cancer_data.target
breast_cancer
```

### 4. Data Preparation

```python
X =pd.DataFrame(data = breast_cancer_data.data,columns = breast_cancer_data.feature_names)
y =  breast_cancer['target'] 
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test  = train_test_split(X,y,train_size=0.8,
                                                     random_state=40,
                                                     shuffle=True,stratify= y)
```

### 5. Model Building

```python
import matplotlib.pyplot as plt

plt.pie(y.value_counts(),autopct = '%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.pie(y_train.value_counts(),autopct = '%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.pie(y_test.value_counts(),autopct = '%1.1f%%')
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier #from sklearn.ensemble import AdaBoostClassifier 
```

```python
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight= {0: 2, 1: 1})
```

### 6. Model Training

```python
dt_model.fit(X_train,y_train)
```

### 7. Model Testing || 8. Model Evaluation

#### Training Data

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

##### Evaluation Metrics

```python
from sklearn.metrics import roc_auc_score,roc_curve,accuracy_score,classification_report,confusion_matrix
```

```python
accuracy_score(y_train,y_pred_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
confusion_matrix(y_train,y_pred_train)
```

```python
print("AUC:",roc_auc_score(y_train,y_pred_train))

fpr,tpr,thresshold = roc_curve(y_train,dt_model.predict_proba(X_train)[:,1])

plt.plot(fpr,tpr,color ='red')
plt.plot([0,1],[0,1],'k--')
```

#### Testing Data

```python
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

##### Evaluation Metrics

```python
accuracy_score(y_test,y_pred_test)
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
confusion_matrix(y_test,y_pred_test)
```

```python
print("AUC:",roc_auc_score(y_test,y_pred_test))

fpr,tpr,thresshold = roc_curve(y_test,dt_model.predict_proba(X_test)[:,1])

plt.plot(fpr,tpr,color ='red')
plt.plot([0,1],[0,1],'k--')
```
