### 2. Data Collection

#### 2.1 Import necessory datsets

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import load_breast_cancer
breast_cancer_data = load_breast_cancer()
```

```python
breast_cancer_df = pd.DataFrame(data = breast_cancer_data.data,columns= breast_cancer_data.feature_names)
breast_cancer_df
```

### 3. Data Understanding

```python
# shape, isna, duplidates, dtypes
```

```python
print(breast_cancer_data.DESCR)
```

### 4. Data Preparation

#### 4.1 Data Preprocessing

```python
std_scaler = StandardScaler()
```

```python
scaled_X = pd.DataFrame(std_scaler.fit_transform(breast_cancer_df),columns= breast_cancer_data.feature_names)
```

```python
scaled_X.describe()
```

#### 4.2 Train Test Split

```python
y = breast_cancer_data.target
```

```python
 X_train, X_test, y_train, y_test =train_test_split(scaled_X,y,test_size =0.2,
                random_state=39,
                shuffle=True,
                stratify=y)
```

### 5. Model Building

```python
import xgboost as xgb
from datetime import datetime
```

```python
xgboost = xgb.XGBClassifier(n_estimators=50,max_depth=3,learning_rate = 0.008,gamma= 1.9)
```

### 6. Model Training

```python
%%time
xgboost.fit(X_train,y_train)
```

### 7. Model Testing 

```python
y_pred_train = xgboost.predict(X_train)
```

```python
y_pred_test = xgboost.predict(X_test)
```

### 8. Model Evalution

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,precision_score,recall_score
```

#### 8.1 Training data Evaluation metrics

```python
accuracy_score(y_train,y_pred_train)
```

```python
precision_score(y_train,y_pred_train)
```

```python
recall_score(y_train,y_pred_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
confusion_matrix(y_train,y_pred_train)
```

#### 8.2 Test data Evaluation metrics

```python
accuracy_score(y_test,y_pred_test)
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
precision_score(y_test,y_pred_test)
```

```python
recall_score(y_test,y_pred_test)
```

```python
confusion_matrix(y_test,y_pred_test)
```
