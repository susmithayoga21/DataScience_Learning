### 2. Data Collection

#### 2.1 Import Nescessory datasets and libiraries

```python
from sklearn.datasets import load_breast_cancer
from sklearn.svm import SVC,SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,mean_absolute_error,root_mean_squared_error

breast_cancer_data = load_breast_cancer()
```

```python
print(breast_cancer_data.DESCR)
```

```python
breast_cancer_data_df = pd.DataFrame(breast_cancer_data.data,columns =breast_cancer_data.feature_names)
```

```python
breast_cancer_data_df["Target"] = breast_cancer_data.target
```

### 4. Data preparation

```python
X = breast_cancer_data_df.drop("Target",axis= 1)
y =breast_cancer_data_df["Target"]
```

```python
X_train, X_test, y_train, y_test =train_test_split(X,y,train_size=0.8,random_state=39,shuffle=True,stratify=y)
```

### 5. Model Building

```python
classification = SVC()
```

```python
classification = SVC(kernel='poly',degree=2)
```

```python
classification = SVC(kernel='linear')
```

### 6. Model Training

```python
%%time
classification.fit(X_train,y_train)
```

### 7. Model Testing

```python
y_pred_train = classification.predict(X_train)
```

```python
y_pred_test = classification.predict(X_test)
```

### 8. Model Evaluation

```python
accuracy_score(y_train,y_pred_train)
```

```python
accuracy_score(y_test,y_pred_test)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
confusion_matrix(y_train,y_pred_train)
```

```python
confusion_matrix(y_test,y_pred_test)
```
