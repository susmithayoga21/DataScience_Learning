### 4. Data processing

```python
X = wine_data_df.drop("Target",axis = 1)
y = wine_data_df["Target"]
```

```python
std_scaler =StandardScaler()

scaled_X = pd.DataFrame(std_scaler.fit_transform(X),columns=wine_data.feature_names)
```

```python
 X_train, X_test, y_train, y_test = train_test_split(scaled_X,y,train_size= 0.8,random_state=39,stratify=y,shuffle=True)
```

### 5. Model Building

```python
knn_model = KNeighborsClassifier()
```

```python
knn_model2 = KNeighborsClassifier(n_neighbors=3)
```

### 6. Model Training

```python
knn_model2.fit(X_train,y_train)
```

### 7. Model Testing

```python
y_pred_train = knn_model.predict(X_train)
```

```python
y_pred_test = knn_model.predict(X_test)
```

```python
y_pred_train2 = knn_model2.predict(X_train)
```

```python
y_pred_test2 = knn_model2.predict(X_test)
```

### 8. Model Evaluation

#### how to find Optimal No of Neighbours

```python
from sklearn.model_selection import KFold,cross_val_score
```

```python
kfold = KFold(n_splits=5,shuffle=True, random_state=30)
cv_score_mean =[]
k_mean_dict = {}
for i in range(3,52,2):
    knn_classifier = KNeighborsClassifier(n_neighbors=i)
    knn_classifier.fit(scaled_X,y)
    cv_score = cross_val_score(estimator=knn_classifier,X= scaled_X,y= y,cv=kfold)
    cv_score_mean.append(cv_score.mean().round(4))
    k_mean_dict.update({i : cv_score.mean().round(4)})
print(k_mean_dict)
    
```
