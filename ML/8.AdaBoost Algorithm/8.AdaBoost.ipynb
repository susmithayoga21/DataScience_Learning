### 2. Data Collection

#### 2.1 Import necessory data sets

```python
from sklearn import datasets 
import pandas as pd
```

```python
wine_data = datasets.load_wine()
```

```python
wine_df = pd.DataFrame(data =wine_data.data,columns=wine_data.feature_names)
wine_df['target'] = wine_data.target
wine_df
```

### 3. Data understanding

```python
wine_df['target_code'] = pd.Categorical.from_codes(wine_data.target, wine_data.target_names)
wine_df
```

### 4. Data Preparation

```python
X = pd.DataFrame(data =wine_data.data,columns=wine_data.feature_names)
y = wine_df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2,random_state=19,shuffle=True)
```

### 5. Model Building

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
classifier = AdaBoostClassifier()
```

### 6. Model Training

```python
classifier.fit(X_train, y_train)
```

### 7. Model Testing || 8.Model Evaluation

##### Training data

```python
y_pred_train = classifier.predict(X_train)
y_pred_train
```

##### Evaluation Metrics

```python
from sklearn.metrics import confusion_matrix, classification_report,accuracy_score,roc_auc_score
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
confusion_matrix(y_train,y_pred_train)
```

```python
accuracy_score(y_train,y_pred_train)
```

```python
y_prob_train = classifier.predict_proba(X_train)
```

```python
roc_auc_score(y_train,y_prob_train,multi_class='ovr')
```

##### Testing data

```python
y_pred_test = classifier.predict(X_test)
y_pred_test
```

##### Evaluation Metrics

```python
print(classification_report(y_test,y_pred_test))
```

```python
accuracy_score(y_test,y_pred_test)
```

```python
confusion_matrix(y_test,y_pred_test)
```
